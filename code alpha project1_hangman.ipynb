```python
import random

def hangman():
    words = ["python", "hangman", "developer", "keyboard", "internship"]
    word = random.choice(words)
    guessed_letters = []
    attempts_left = 6

    hangman_stages = [
        "",
        "😵",
        "😵💦",
        "😵💦💦",
        "😰",
        "😰❌",
        "💀 GAME OVER"
    ]

    print("Welcome to Hangman!")
    print(f"You have {attempts_left} incorrect guesses allowed.")

    while attempts_left > 0:
        display = ""
        for letter in word:
            if letter in guessed_letters:
                display += letter
            else:
                display += "_"
        print("\nWord: " + display)
        print(f"Guessed letters: {', '.join(guessed_letters) if guessed_letters else 'None yet'}")
        print(hangman_stages[6 - attempts_left])

        if display == word:
            print("You guessed the word! You win!")
            break

        guess = input("Guess a letter: ").lower()

        if guess in guessed_letters:
            print("You already guessed that letter.")
        elif guess in word:
            guessed_letters.append(guess)
            print("Correct!")
        else:
            guessed_letters.append(guess)
            attempts_left -= 1
            print(f"Wrong! You have {attempts_left} attempts left.")

    if attempts_left == 0:
        print(f"\nGame over! The word was: {word}")

    play_again = input("\nPlay again? (y/n): ").lower()
    if play_again == "y":
        hangman()
    else:
        print("Thanks for playing!")

hangman()
```
